```python
"""
Exploratory Data Analysis script for notebook export.
Generates correlation matrices and temporal commuter signature plots.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.feature_engineering import engineer_all_features


def run_eda_pipeline(data_path: str, output_img_dir: str = "reports/figures"):
    import os
    os.makedirs(output_img_dir, exist_ok=True)
    
    # 1. Load Data
    raw_df = pd.read_csv(data_path, encoding="unicode_escape")
    df = engineer_all_features(raw_df)
    
    # 2. Set Plotting Style
    sns.set_theme(style="whitegrid", font_scale=1.1)
    
    # --- Visualization 1: Commuter Demand Profile ---
    plt.figure(figsize=(10, 5))
    sns.lineplot(
        data=df,
        x="Hour",
        y="Rented Bike Count",
        hue="IsWeekend",
        palette={0: "#1f77b4", 1: "#ff7f0e"},
        marker="o",
        linewidth=2.5,
    )
    plt.title("Commuter Hourly Rental Demand Profile (Weekday vs. Weekend)", fontsize=14, fontweight="bold")
    plt.xlabel("Hour of Day (0-23)")
    plt.ylabel("Mean Rented Bike Count")
    plt.legend(title="Day Type", labels=["Weekday", "Weekend"])
    plt.tight_layout()
    plt.savefig(os.path.join(output_img_dir, "commuter_hourly_profile.png"), dpi=300)
    plt.close()
    
    # --- Visualization 2: Correlation Heatmap ---
    plt.figure(figsize=(12, 8))
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    corr_matrix = df[numeric_cols].corr()
    
    sns.heatmap(
        corr_matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        square=True,
        cbar_kws={"shrink": 0.8},
    )
    plt.title("Feature Correlation Matrix (Weather & Engineered Metrics)", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig(os.path.join(output_img_dir, "correlation_heatmap.png"), dpi=300)
    plt.close()
    
    print(f"EDA plots generated successfully in '{output_img_dir}/'!")


if __name__ == "__main__":
    run_eda_pipeline("data/raw/SeoulBikeData.csv")
```
